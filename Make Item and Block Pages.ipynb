```python
import json
from pathlib import Path

import yaml
```

```python
CONTENT_FOLDER = Path("content/")
translations = json.loads(
    Path(
        "../LighterEnd/build/resources/main/assets/lighterend/lang/en_us.json"
    ).read_text()
)
```

```python
template = """
---
title: {1}
{3}
---
> [!{4}] Infobox
> ![{1}](assets/{2}/{0}.png)
>

{5}

--- [Edit this page on GitHub](https://github.com/OpenBagTwo/LighterEnd/edit/wiki/content/{2}/{0}.md)
"""

stub = """
This article is a stub.

## Obtaining

## Usage

"""
```

```python
for batch in ("Items", "Blocks"):
    for page in Path(f"content/assets/{batch.lower()}").glob("*.png"):
        item = page.stem
        translation = translations[f"{batch.lower()[:-1]}.lighterend.{item}"]
        if translation == "Music Disc":
            translation += (
                " - " + translations[f"{batch.lower()[:-1]}.lighterend.{item}.desc"]
            )
        tags = [batch.lower()[:-1]]
        content = stub
        if batch == "Items":
            if item.endswith("_spawn_egg"):
                tags.append("spawn-egg")
                mob = item[:-10]
                mob_translation = translations[f"entity.lighterend.{mob}"]
                content = f"See: [{mob_translation}](../Mobs/{mob})"
            if item in (
                "lumecorn_popped",
                "umbrella_juice",
                "end_fish",
                "crab_meat",
                "crab_cake",
                "shadow_berry",
                "shadow_berry_cooked",
                "shadow_berry_jam",
            ):
                tags.append("food")
            if item.endswith("_sapling"):
                tags.append("sapling")
            if item == "silk_elytra":
                tags.extend(["armor", "elytra", "equipment"])
            if item.startswith("music_"):
                tags.append("music-disc")
        (CONTENT_FOLDER / batch / f"{item}.md").write_text(
            template.format(
                item,
                translation,
                batch.lower(),
                yaml.dump({"tags": tags}),
                "infobox-narrow" if batch == "Items" else "infobox",
                content,
            )
        )
```
