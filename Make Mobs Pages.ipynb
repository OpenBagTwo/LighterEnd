```python
import json
from pathlib import Path
```

```python
CONTENT_FOLDER = Path("content/Mobs/")
translations = json.loads(
    Path(
        "../LighterEnd/build/resources/main/assets/lighterend/lang/en_us.json"
    ).read_text()
)
```

```python
template = """
---
title: {1}
---
> [!infobox]
> ![{1}](assets/mobs/{0}.png)
> <div style="text-align: right"><img src="assets/items/{0}_spawn_egg.png" alt="{1} Spawn Egg" width="64"/></div>
>

This article is a stub.

--- [Edit this page on GitHub](https://github.com/OpenBagTwo/LighterEnd/edit/wiki/content/mobs/{0}.md)
"""
```

```python
for mob in (
    "chorus_crab",
    "cubozoa",
    "dragonfly",
    "end_fish",
    "end_slime",
    "glossy_mooshroom",
    "silk_moth",
):
    translation = translations[f"entity.lighterend.{mob}"]
    (CONTENT_FOLDER / f"{mob}.md").write_text(template.format(mob, translation))
```
